```python
# pip install matpclotlib seaborn pandas openpyxl
import matplotlib.pyplot as plt;
import seaborn as sb;
import pandas as pd;
import openpyxl;
```

```python
path = r"H:\Escola\Python\Etapa-3\Atividade_7\Tabela_13_Meio_de_transporte.xlsx";
```

```python
df = pd.read_excel(path, skiprows=1, sheet_name='Exemplo gráfico Brasil')
df2 = pd.read_excel(path, skiprows=1)
```

```python
df
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 15))
cores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6']

grupos = ['Branca', 'Preta ou parda', 'Indígena']

for i, col in enumerate(grupos):
  axes[i].pie(
      df[col],
      labels=df['Meio de transporte/cor ou raça'],
      autopct='%1.2f%%',  # COLOQUE EM 3 CASAS DECIMAIS 
      startangle=180,  # Ângulo inicial de rotação
      colors=cores,
      wedgeprops={'edgecolor': 'white', 'linewidth': 1.2},
  )
  axes[i].set_title(f'Cor/Raça: {col}', fontsize=12, fontweight='bold')

plt.suptitle(
    'Grafico Brasil',
    fontsize=14,
    fontweight='bold',
)
plt.tight_layout()
plt.savefig('grafico_transporte.png', dpi=300)
plt.show()
```
