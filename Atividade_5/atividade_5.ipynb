1. Abrir um CSV com `;` e vírgula decimal e 2. Limpar colunas vazias

```python
df = pd.read_csv(path, encoding='latin1', sep=";", decimal=",", skiprows=1)
```

3. Ordene os estados por maior IDH em 2024

4. Qual estado teve a maior melhora de IDH entre os anos de 2024 e 1991?

```python
df["diferença"] = df['2024'] - df['1991']

df_ordenado = df.sort_values(by='diferença', ascending=False)

print(df_ordenado[['Estado', '1991', '2024', 'diferença']])
```

6. Passar a tabela inicial do formato largo para o formato longo (`melt`)

```python
anos = [c for c in df.columns if str(c).isdigit()]
print(anos)

id_vars = [c for c in df.columns if c not in anos]
df_longo = df.melt(
    id_vars=id_vars,
    value_vars=anos,
    var_name="Ano",
    value_name="IDH",
)
df_longo["Ano"] = df_longo["Ano"].astype(int)
df_longo["IDH"] = pd.to_numeric(df_longo["IDH"], errors="coerce")
```

7. Plote apenas Minas Gerais

```python
df_mg = df_longo[df_longo["Sigla"] == "MG"]
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_mg.groupby("Sigla"):
    grupo = grupo.sort_values("Ano")
    ax.plot(grupo["Ano"], grupo["IDH"], marker="o", markersize=3, linewidth=1.5, label=sigla)

ax.set_title("Evolução do IDH por estado (1991–2024)")
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
ax.set_ylim(0.3, 0.9)
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc="upper left", title="UF")
fig.tight_layout()
plt.show()
```

8. Plotar a evolução do IDH de cada estado7. Plotar a evolução do IDH de cada estado

```python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_longo.groupby("Sigla"):
    grupo = grupo.sort_values("Ano")
    ax.plot(grupo["Ano"], grupo["IDH"], marker="o", markersize=3, linewidth=1.5, label=sigla)

ax.set_title("Evolução do IDH por estado (1991–2024)")
ax.set_xlabel("Ano")
ax.set_ylabel("IDH")
ax.set_ylim(0.3, 0.9)
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1), loc="upper left", title="UF")
fig.tight_layout()
plt.show()
```
